### Set Matrix Zeroes

Input: matrix=[[1,1,1],[1,0,1],[1,1,1]]
Output: [[1,0,1],[0,0,0],[1,0,1]]
Explanation: Since matrix[2][2]=0.Therfore the 2nd column and 2nd row wil be set to 0.

Input: matrix=[[0,1,2,0],[3,4,5,2],[1,3,1,5]]
Output:[[0,0,0,0],[0,4,5,0],[0,3,1,0]]
Explanation:Since matrix[0][0]=0 and matrix[0][3]=0. Therefore 1st row, 1st column and 4th column will be set to 0

```python
# Brute Force approach

def set_matrix_zeros_brute_force(matrix):
    number_of_rows = len(matrix)
    number_of_columns = len(matrix[0])
    # for loop to traverse the whole matrix and 
    # identify the zeros in it then
    # set those columns and rows to -1
    for i in range(number_of_rows):
        for j in range(number_of_columns):
            if matrix[i][j] == 0:
                for r in range(number_of_rows):
                    if matrix[r][j] != 0:
                        matrix[r][j] = -1
                for c in range(number_of_columns):
                    if matrix[i][c] != 0:
                        matrix[i][c] = -1
    # for loop to convert the -1 to 0
    for i in range(number_of_rows):
        for j in range(number_of_columns):
            if matrix[i][j] == -1:
                matrix[i][j] = 0
    return matrix

# Driver code
"""Usually given"""
matrix = [[[1,1,1],[1,0,1],[1,1,1]],[[0,1,2,0],[3,4,5,2],[1,3,1,5]]]
solution = [set_matrix_zeros_brute_force(matrix[0]), set_matrix_zeros_brute_force(matrix[1])]
for s in solution:
    print(s)
```

```python
%%script node
// use %%script node for Node.js/Server-Side Execution
// use %%javascript or %%js for browser based JS execution

// Brute force solution 
function setMatrixZeros(matrix){
    const m = matrix.length
    const n = matrix[0].length
    for(let i = 0; i < m; i++){
        for(let j=0;j<n;j++){
            if(matrix[i][j] === 0){
                for(let r=0;r<m;r++){
                    if(matrix[r][j] != 0){
                        matrix[r][j] = -1;
                    }
                }
                for(let c=0;c<n;c++){
                    if(matrix[i][c]!=0){
                        matrix[i][c] = -1;
                    }
                }
            }
        }
    }
    for(let i=0;i<m;i++){
        for(let j=0;j<n;j++){
            if(matrix[i][j] === -1){
                matrix[i][j]=0;
            }
        }
    }
    return matrix;
}
// Driver Code
/*Usually Given*/
let matrix = [[[1,1,1],[1,0,1],[1,1,1]],[[0,1,2,0],[3,4,5,2],[1,3,1,5]]];
let sol = [setMatrixZeros(matrix[0]), setMatrixZeros(matrix[1])];
for (let i=0;i<sol.length;i++){
    console.log(sol[i]);
}
```

```python
%%script node

// Better Approach
class SolutionBetter{
    setMatrixZerosBetter(matrix){
        const m = matrix.length;
        const n = matrix[0].length;
        // creating row and col markers
        let rowMarker = new Array(m).fill(false);
        let colMarker = new Array(n).fill(false);
        // traversing the matrix 
        for(let i=0; i<m; i++){
            for(let j=0; j<n; j++){
                // filling the row and col markers 
                if(matrix[i][j]===0){
                    rowMarker[i] = true;
                    colMarker[j] = true;
                }
            }
        }
        // using the marker for zeroing the matrix
        for(let i=0; i<m; i++){
            for(let j=0; j<n; j++){
                if(rowMarker[i]===true||colMarker[j]===true){
                    matrix[i][j]=0;
                }
            }
        }
        return matrix
    }
}


// Driver Code
/*Usually Given*/
let matrix = [[[1,1,1],[1,0,1],[1,1,1]],[[0,1,2,0],[3,4,5,2],[1,3,1,5]]];
let sol = new SolutionBetter()
let solutionMatrix = [sol.setMatrixZerosBetter(matrix[0]), sol.setMatrixZerosBetter(matrix[1])];
for (let i=0;i<solutionMatrix.length;i++){
    console.log(solutionMatrix[i]);
}
```

```python
# Optimal Approach
class OptimalSolution:
    def set_matrix_zero_optimal(self, matrix:list[list[int]])->list[list[int]]:
        firstRowZero = False
        firstColZero = False
        r = len(matrix)
        c = len(matrix[0])
        # check if first row has any zero
        for j in range(c):
            if matrix[0][j] == 0:
                firstRowZero=True
                break
        # check if first col has any zero
        for i in range(r):
            if matrix[i][0]==0:
                firstColZero = True
        # use the first row and first col as markers
        for i in range(1, r):
            for j in range(1, c):
                if matrix[i][j]==0:
                    matrix[i][0] = 0
                    matrix[0][j] = 0
        # zeroing based on those markers
        for i in range(1,r):
            for j in range(1,c):
                if matrix[i][0]==0 or matrix[0][j]==0:
                    matrix[i][j]=0 
        # setting first row to zero if true
        if firstRowZero == True:
            for j in range(c):
                matrix[0][j]=0
        # setting first col to zero if true
        if firstColZero == True:
            for i in range(r):
                matrix[i][0]=0  
        return matrix

# Driver Code
matrix = [[[1,0,1],[1,1,1],[1,1,1]],[[0,1,2,0],[3,4,5,2],[1,3,1,5]]]
sol = OptimalSolution()
solution = [sol.set_matrix_zero_optimal(matrix[0]),sol.set_matrix_zero_optimal(matrix[1])]
for s in solution:
    print(s)
```

```python
%%script node

// Optimal Approach
class SolutionOptimal{
    setMatrixZerosOptimal(matrix){
        const m = matrix.length;
        const n = matrix[0].length;
        // flags to mark the first row and col
        let firstRowZero = false;
        let firstColZero = false;
        // checking if first row and cols have zero or not
        for(let j=0; j<n;j++){
            if(matrix[0][j]===0){
                firstRowZero = true;
                break;
            }
        }
        for(let i=0; i<m;i++){
            if(matrix[i][0]===0){
                firstColZero = true;
                break;
            }
        }
        // then using first row and col to mark other zeros
        for(let i=1;i<m;i++){
            for(let j=1;j<n;j++){
                if(matrix[i][j]===0){
                    matrix[0][j]=0;
                    matrix[i][0]=0;
                }
            }
        }
        // then zeroing based on those markers
        for(let i=1;i<m;i++){
            for(let j=1;j<n;j++){
                if(matrix[0][j]===0||matrix[i][0]===0){
                    matrix[i][j]=0;
                }
            }
        }
        // now converting the first row and col to zero
        if(firstRowZero===true){
            for(let j=0;j<n;j++){
                matrix[0][j]=0
            }
        }
        if(firstColZero===true){
            for(let i=0;i<m;i++){
                matrix[i][0]=0
            }
        }
        return matrix
    }
}

// Driver Code
/*Usually Given*/
let matrix = [[[1,1,1],[1,0,1],[1,1,1]],[[0,1,2,0],[3,4,5,2],[1,3,1,5]]];
let sol = new SolutionOptimal()
let solutionMatrixOp = [sol.setMatrixZerosOptimal(matrix[0]), sol.setMatrixZerosOptimal(matrix[1])];
for (let i=0;i<solutionMatrixOp.length;i++){
    console.log(solutionMatrixOp[i]);
}
```
